```python
from fairdatanow import DataViewer
import os
import warnings
#warnings.filterwarnings(action='once')
warnings.filterwarnings('ignore')
```

```python
configuration = {
    'url': "https://laboppad.nl/falnama-project", 
    'user':    os.getenv('NC_AUTH_USER'),
    'password': os.getenv('NC_AUTH_PASS')
}
```

```python
filters = {'columns': ['path', 'size', 'modified'],
 'extensions': ['.tif'],
 'search': r'.*grenz.*\.TIF$',
 'show_directories': False,
 'show_filters': True,
 'use_regex': True}
```

```python
dv = DataViewer(configuration, **filters)
dv
```

```python
files = dv.download_filtered()
```

```python
from matplotlib import pyplot as plt

fig, axs = plt.subplots(nrows=7,ncols=4)
axs = axs.flatten()
for ax, path in zip(axs,files):
    img = plt.imread(path)
    ax.imshow(img)
```
